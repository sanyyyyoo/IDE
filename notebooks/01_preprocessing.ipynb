```python
# notebooks/01_preprocessing.ipynb

import pandas as pd
import os

# Paths
RAW_PATH = "../data/processed/cleaned.csv"  # <-- changed to cleaned.csv
PROCESSED_PATH = "../data/processed/labeled.csv"

# 1. Load raw data
df = pd.read_csv(RAW_PATH)  # <-- changed to read_csv
print("Raw data shape:", df.shape)
print("Columns:", df.columns.tolist())
df.head()

# 2. Rename columns consistently
# (Change these names if your CSV columns differ)
df = df.rename(columns={
    "Business Name": "name",
    "Business Category": "category",
    "Address": "address",
    "Contact": "phone"
})

# 3. Drop duplicates and empty rows
df = df.drop_duplicates().dropna(how="all")

# 4. Convert to "long format" for classification training
#    Each row → one text snippet with its label
records = []

for _, row in df.iterrows():
    if pd.notna(row["name"]):
        records.append({"text": row["name"], "label": "name"})
    if pd.notna(row["category"]):
        records.append({"text": row["category"], "label": "category"})
    if pd.notna(row["address"]):
        records.append({"text": row["address"], "label": "address"})
    if pd.notna(row["phone"]):
        records.append({"text": str(row["phone"]), "label": "phone"})

# 5. Create training DataFrame
labeled_df = pd.DataFrame(records)
print("Processed dataset shape:", labeled_df.shape)
print(labeled_df["label"].value_counts())
labeled_df.head(10)

# 6. Save for training
os.makedirs("../data/processed", exist_ok=True)
labeled_df.to_csv(PROCESSED_PATH, index=False)
print("✅ Saved processed data to:", PROCESSED_PATH)
```

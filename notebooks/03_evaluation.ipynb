```python
# notebooks/03_testing.ipynb
import pandas as pd
import torch
from transformers import pipeline

# Load model
classifier = pipeline(
    "text-classification",
    model="../models/distilbert_finetuned",
    device=0 if torch.cuda.is_available() else -1
)

# Test samples
test_samples = [
    "Deepa Sales / Nikhil Designer Studio, Shop no-4, Chitalsar, circle, Tikuji Ni Wadi Rd, near Manpada",
    "Kajaria Galaxy Showroom",
    "+918928047473"

]

print("🔍 Model Predictions:")
for text in test_samples:
    result = classifier(text)
    print(f"'{text}' → {result[0]['label']} ({result[0]['score']:.3f})")
```
